## 1.1 Variables et types
- Variable = **référence** vers un objet. Types : `int`, `float`, `bool`, `str`.
- ⚠️ Piège : `=` (affectation) ≠ `==` (test).
- 💡 `int` en précision arbitraire : pas de dépassement.

## 1.2 Boucles for / while
- `for` : nombre de tours connu. `while` : arrêt dépend d'une condition.
- ⚠️ `while` : n'oublie pas de mettre à jour la variable de contrôle (boucle infinie).
- 💡 Méthode while : variable de contrôle → valeur u_k → inéquation d'arrêt → nombre de tours.

```python
# for avec range
sum_val = 0
for i in range(1, 5):
    sum_val += i
assert sum_val == 10

# while logarithmique : combien de tours ?
i, tours = 1, 0
while i <= 16:
    i = i * 2
    tours += 1
print("tours =", tours)   # 5 = log2(16) + 1
assert tours == 5
```

## 1.3 Fonctions et portée (scope)
- ⚠️ Oublier `return` → la fonction renvoie `None`.
- 💡 Une fonction s'arrête au **premier** `return`.

```python
def carre(x):
    res = x * x      # res est locale
    return res

def oubli(x):
    res = x * x      # pas de return !

assert carre(5) == 25
assert oubli(5) is None
```

# PARTIE 2 — Structures de données

## Tableau des coûts (liste Python)
| Opération | Complexité |
|---|---|
| `L[i]`, `len(L)` | O(1) |
| `L.append(x)`, `L.pop()` | O(1) amorti |
| `L.insert(0,x)`, `L.pop(0)`, `del L[0]` | **O(n)** ⚠️ |
| `x in L`, `min`, `max`, `sum` | O(n) |
| Slicing `L[i:j]`, `L.copy()` | O(j−i) / O(n) |
| `L1 + L2` | O(n1+n2) |
| `sorted(L)`, `L.sort()` | O(n log n) |

```python
import time
n = 50_000
L = list(range(n))

t = time.time()
for _ in range(n // 5): L.pop()      # fin : O(1)
t_fin = time.time() - t

L = list(range(n))
t = time.time()
for _ in range(n // 5): L.pop(0)     # début : O(n)
t_debut = time.time() - t

print(f"pop() fin   : {t_fin:.4f}s")
print(f"pop(0) début: {t_debut:.4f}s")
```

## 2.0 Concaténation : `+`, `+=`, `append`, `extend`, `join`
- `L1 + L2` crée une **nouvelle** liste (O(n1+n2)).
- `append(x)` ajoute **un** élément ; `extend(L2)` ajoute **tous** les éléments de L2.
- Construire une chaîne : **`''.join(liste)`** et pas `s += ...` dans une boucle.

```python
L = [1, 2]
L.append([3, 4]);  assert L == [1, 2, [3, 4]]   # un seul élément (une liste)
L = [1, 2]
L.extend([3, 4]);  assert L == [1, 2, 3, 4]

mots = ["a", "b", "c"]
assert "-".join(mots) == "a-b-c"

# += sur liste modifie en place, + crée une nouvelle liste
A = [1]; B = A
A += [2]
assert B == [1, 2]       # B a changé (même objet)
A = [1]; B = A
A = A + [2]
assert B == [1]          # B inchangé (A est un nouvel objet)
```

## 2.1 Aliasing, copie superficielle vs profonde
- `B = A` → **même objet** (aliasing). `A.copy()` / `A[:]` → copie **superficielle**.
- Pour listes de listes : `copy.deepcopy`.
- `A is B` teste l'identité, `A == B` teste l'égalité des valeurs.

```python
A = [1, 2, 3]
B = A            # aliasing
C = A.copy()     # copie
B[0] = 99
assert A[0] == 99 and C[0] == 1
assert A is B and A is not C

# Piège classique : matrice mal construite
M_faux = [[0] * 3] * 3          # 3 fois la MÊME ligne
M_faux[0][0] = 1
assert M_faux[1][0] == 1        # ⚠️ toutes les lignes changent !

M_ok = [[0] * 3 for _ in range(3)]
M_ok[0][0] = 1
assert M_ok[1][0] == 0

# Copie superficielle vs profonde
import copy
X = [[1, 2], [3, 4]]
Y = X.copy()
Z = copy.deepcopy(X)
X[0][0] = 100
assert Y[0][0] == 100     # sous-listes partagées
assert Z[0][0] == 1       # deepcopy indépendante
```

## 2.2 Pile (LIFO) et File (FIFO)
- Pile : `append` / `pop`. File : `deque` avec `append` / `popleft`.
- ⚠️ Pas de `L.pop(0)` dans une boucle (O(n²)).

```python
from collections import deque

pile = []
pile.append(10); pile.append(20)
assert pile.pop() == 20          # dernier entré, premier sorti

file = deque()
file.append(10); file.append(20)
assert file.popleft() == 10      # premier entré, premier sorti

# Application : parenthésage
def bien_parenthese(s):
    pile = []
    paires = {')': '(', ']': '[', '}': '{'}
    for c in s:
        if c in '([{':
            pile.append(c)
        elif c in paires:
            if not pile or pile.pop() != paires[c]:
                return False
    return not pile

assert bien_parenthese("([]{})")
assert not bien_parenthese("([)]")
assert not bien_parenthese("((")
```

# PARTIE 3 — Récursivité

## 3.1 Principe
Une fonction récursive a : **un cas de base** + **un appel sur un problème plus petit**.
⚠️ Chaque appel en attente occupe un cadre de pile → `RecursionError` si trop profond.

```python
def factorielle(n):
    if n == 0:                    # cas de base
        return 1
    return n * factorielle(n - 1)

assert factorielle(5) == 120

import sys
print("limite de récursion :", sys.getrecursionlimit())

def inf(n):            # pas de cas de base !
    return inf(n + 1)
try:
    inf(0)
except RecursionError:
    print("RecursionError attrapée")
```

### Récurrences à connaître
| T(n) | Complexité | Exemple |
|---|---|---|
| T(n−1)+O(1) | O(n) | factorielle |
| T(n−1)+O(n) | O(n²) | tri sélection récursif |
| T(n/2)+O(1) | O(log n) | dichotomie, expo rapide |
| 2T(n/2)+O(1) | O(n) | max par dichotomie |
| 2T(n/2)+O(n) | O(n log n) | tri fusion |
| 2T(n−1)+O(1) | O(2ⁿ) | Hanoï |
| T(n−1)+T(n−2)+O(1) | O(φⁿ) | Fibonacci naïf |

## 3.2 Exponentiation rapide

```python
def puissance_rec(x, n):          # O(log n) temps, O(log n) espace
    if n == 0: return 1
    y = puissance_rec(x, n // 2)  # UN SEUL appel, mémorisé dans y
    return y * y if n % 2 == 0 else y * y * x

def puissance_iter(x, n):         # O(log n) temps, O(1) espace
    r = 1
    while n > 0:
        if n % 2 == 1: r = r * x
        x = x * x
        n = n // 2
    return r

for x in (2, 3, 5):
    for n in range(0, 20):
        assert puissance_rec(x, n) == x**n == puissance_iter(x, n)
```

## 3.3 Autres classiques : PGCD, Fibonacci + mémoïsation, Hanoï

```python
def pgcd(a, b):
    return a if b == 0 else pgcd(b, a % b)
assert pgcd(48, 18) == 6

def fib_naif(n):                 # O(φⁿ) : exponentiel !
    return n if n < 2 else fib_naif(n-1) + fib_naif(n-2)

def fib_memo(n, memo=None):      # O(n)
    if memo is None: memo = {}
    if n < 2: return n
    if n not in memo:
        memo[n] = fib_memo(n-1, memo) + fib_memo(n-2, memo)
    return memo[n]

assert fib_naif(15) == fib_memo(15) == 610
assert fib_memo(90) == 2880067194370816120

def hanoi(n, a, b, c, coups):    # déplacer n disques de a vers c
    if n == 0: return
    hanoi(n-1, a, c, b, coups)
    coups.append((a, c))
    hanoi(n-1, b, a, c, coups)

coups = []; hanoi(4, 'A', 'B', 'C', coups)
assert len(coups) == 2**4 - 1
```

# PARTIE 4 — Algorithmique

## 4.1 Dichotomie (liste **triée** uniquement)
⚠️ `while g <= d` (pas `<`), et `d = m - 1` / `g = m + 1`. Complexité O(log n).

```python
def dichotomie(x, L):
    g, d = 0, len(L) - 1
    while g <= d:
        m = (g + d) // 2
        if L[m] == x: return m
        elif L[m] < x: g = m + 1
        else: d = m - 1
    return None

L = [1, 3, 5, 7, 9, 11]
for i, v in enumerate(L):
    assert dichotomie(v, L) == i
assert dichotomie(4, L) is None
assert dichotomie(1, [1]) == 0
```

## 4.2 Tris par sélection et insertion (O(n²))

```python
def tri_selection(L):
    n = len(L)
    for i in range(n - 1):
        imin = i
        for j in range(i + 1, n):
            if L[j] < L[imin]: imin = j
        L[i], L[imin] = L[imin], L[i]

def tri_insertion(L):             # meilleur cas O(n) si déjà trié
    for i in range(1, len(L)):
        x = L[i]; j = i - 1
        while j >= 0 and L[j] > x:
            L[j + 1] = L[j]; j -= 1
        L[j + 1] = x

import random
for tri in (tri_selection, tri_insertion):
    for _ in range(100):
        L = [random.randint(0, 50) for _ in range(random.randint(0, 20))]
        M = L.copy(); tri(M)
        assert M == sorted(L)
```

## 4.3 Tri fusion (O(n log n) dans tous les cas)
Stable grâce au `<=`. ⚠️ Les slicings `L[:m]` copient (O(n)) ; mémoire auxiliaire O(n).

```python
def fusion(A, B):
    res = []; i = j = 0
    while i < len(A) and j < len(B):
        if A[i] <= B[j]: res.append(A[i]); i += 1
        else:            res.append(B[j]); j += 1
    res.extend(A[i:]); res.extend(B[j:])
    return res

def tri_fusion(L):
    if len(L) <= 1: return L
    m = len(L) // 2
    return fusion(tri_fusion(L[:m]), tri_fusion(L[m:]))

for _ in range(100):
    L = [random.randint(0, 50) for _ in range(random.randint(0, 30))]
    assert tri_fusion(L) == sorted(L)
```

## 4.4 Comparaison de vitesse (pour sentir les complexités)

```python
import time
def chrono(f, L):
    M = L.copy(); t = time.time(); f(M); return time.time() - t

L = [random.random() for _ in range(3000)]
print(f"insertion : {chrono(tri_insertion, L):.3f}s")
print(f"fusion    : {chrono(lambda M: tri_fusion(M), L):.3f}s")
print(f"sorted    : {chrono(sorted, L):.5f}s")
```

# PARTIE 6 — Réflexes concours

| Si le sujet demande… | J'utilise… |
|---|---|
| xⁿ ou puissance de matrice | Exponentiation rapide O(log n) |
| Chercher dans liste triée | Dichotomie |
| Trier avec garantie | Tri fusion |
| Évaluer un polynôme | **Horner** |
| Historique / parenthésage | Pile |
| Justifier complexité récursive | Écrire T(n) puis résoudre |

```python
def horner(coeffs, x):
    # P(x) = a0 + a1 x + ... + an x^n ; coeffs = [a0, a1, ..., an]
    r = 0
    for a in reversed(coeffs):
        r = r * x + a
    return r

assert horner([1, 2, 3], 2) == 1 + 2*2 + 3*4   # = 17
```
