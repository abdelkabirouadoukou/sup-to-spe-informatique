## Exercice 1 — Somme des chiffres (récursif)

Écris `somme_chiffres(n)` qui renvoie la somme des chiffres d'un entier positif, **sans conversion en `str`**.

*Indice : `n % 10` donne le dernier chiffre, `n // 10` l'enlève.*


````python
def somme_chiffres(n):
    pass   # TODO

# assert somme_chiffres(1234) == 10
# assert somme_chiffres(9) == 9
````

**Corrigé.**

````python
def somme_chiffres(n):
    return n if n < 10 else n % 10 + somme_chiffres(n // 10)

assert somme_chiffres(1234) == 10
assert somme_chiffres(9) == 9
assert somme_chiffres(1000) == 1
````

## Exercice 2 — Renverser sans modifier l'original

Écris `renverse(L)` qui renvoie une **nouvelle** liste inversée ; `L` doit rester intacte.


````python
def renverse(L):
    pass   # TODO

# L = [1, 2, 3]
# assert renverse(L) == [3, 2, 1] and L == [1, 2, 3]
````

**Corrigé.**

````python
def renverse(L):
    return L[::-1]

L = [1, 2, 3]
assert renverse(L) == [3, 2, 1] and L == [1, 2, 3]
assert renverse([]) == []
````

## Exercice 3 — Liste triée ?

Écris `est_triee(L)` qui renvoie `True` si la liste est croissante (au sens large).


````python
def est_triee(L):
    pass   # TODO

# assert est_triee([1, 2, 2, 5]) and not est_triee([2, 1])
````

**Corrigé.**

````python
def est_triee(L):
    return all(L[i] <= L[i + 1] for i in range(len(L) - 1))

assert est_triee([1, 2, 2, 5])
assert not est_triee([2, 1])
assert est_triee([]) and est_triee([7])
````

## Exercice 4 — Doublons ?

Écris `a_doublon(L)` qui renvoie `True` s'il existe au moins deux éléments égaux.


````python
def a_doublon(L):
    pass   # TODO

# assert a_doublon([1, 2, 1]) and not a_doublon([1, 2, 3])
````

**Corrigé.**

````python
def a_doublon(L):
    return len(set(L)) != len(L)

assert a_doublon([1, 2, 1])
assert not a_doublon([1, 2, 3])
assert not a_doublon([])
````

## Exercice 5 — Compter les occurrences

Écris `compter_occurrences(L)` qui renvoie un dictionnaire `{élément: nombre}`.


````python
def compter_occurrences(L):
    pass   # TODO

# assert compter_occurrences("abca") == {"a": 2, "b": 1, "c": 1}
````

**Corrigé.**

````python
def compter_occurrences(L):
    compteur = {}
    for x in L:
        compteur[x] = compteur.get(x, 0) + 1
    return compteur

assert compter_occurrences("abca") == {"a": 2, "b": 1, "c": 1}
assert compter_occurrences([1, 1, 1]) == {1: 3}
````

## Exercice 6 — Maximum par dichotomie

Écris `maximum(L, g, d)` qui renvoie le maximum de `L[g..d]` par division récursive (T(n) = 2T(n/2) + O(1)).


````python
def maximum(L, g=0, d=None):
    pass   # TODO

# assert maximum([3, 9, 2, 7]) == 9
````

**Corrigé.**

````python
def maximum(L, g=0, d=None):
    if d is None:
        d = len(L) - 1
    if g == d:
        return L[g]
    m = (g + d) // 2
    return max(maximum(L, g, m), maximum(L, m + 1, d))

assert maximum([3, 9, 2, 7]) == 9
assert maximum([5]) == 5
assert maximum([1, 2, 3, 4]) == 4
````

## Exercice 7 — Exponentiation rapide (itérative)

Écris `puissance(x, n)` en O(log n) temps et O(1) espace, sans `**`.


````python
def puissance(x, n):
    pass   # TODO

# assert puissance(2, 10) == 1024
# assert puissance(3, 0) == 1
````

**Corrigé.**

````python
def puissance(x, n):
    resultat = 1
    while n > 0:
        if n % 2 == 1:
            resultat *= x
        x *= x
        n //= 2
    return resultat

assert puissance(2, 10) == 1024
assert puissance(3, 0) == 1
for x in (2, 5):
    for n in range(15):
        assert puissance(x, n) == x**n
````

## Exercice 8 — Recherche dichotomique

Écris `dichotomie(x, L)` qui renvoie l'indice de `x` dans la liste **triée** `L`, ou `None`.


````python
def dichotomie(x, L):
    pass   # TODO

# L = [1, 3, 5, 7, 9]
# assert dichotomie(7, L) == 3 and dichotomie(4, L) is None
````

**Corrigé.**

````python
def dichotomie(x, L):
    g, d = 0, len(L) - 1
    while g <= d:
        m = (g + d) // 2
        if L[m] == x:
            return m
        elif L[m] < x:
            g = m + 1
        else:
            d = m - 1
    return None

L = [1, 3, 5, 7, 9]
assert dichotomie(7, L) == 3
assert dichotomie(1, L) == 0
assert dichotomie(9, L) == 4
assert dichotomie(4, L) is None
assert dichotomie(1, []) is None
````

## Exercice 9 — Fusion de deux listes triées

Écris `fusion(A, B)` qui renvoie une liste triée fusionnant `A` et `B` (déjà triées), en O(len(A)+len(B)).


````python
def fusion(A, B):
    pass   # TODO

# assert fusion([1, 3, 5], [2, 4]) == [1, 2, 3, 4, 5]
````

**Corrigé.**

````python
def fusion(A, B):
    res = []
    i = j = 0
    while i < len(A) and j < len(B):
        if A[i] <= B[j]:
            res.append(A[i])
            i += 1
        else:
            res.append(B[j])
            j += 1
    res.extend(A[i:])
    res.extend(B[j:])
    return res

assert fusion([1, 3, 5], [2, 4]) == [1, 2, 3, 4, 5]
assert fusion([], [1, 2]) == [1, 2]
assert fusion([1], []) == [1]
````

## Exercice 10 — Crible d'Ératosthène

Écris `crible(n)` qui renvoie la liste des nombres premiers ≤ n, en O(n log log n).


````python
def crible(n):
    pass   # TODO

# assert crible(20) == [2, 3, 5, 7, 11, 13, 17, 19]
````

**Corrigé.**

````python
def crible(n):
    if n < 2:
        return []
    est_premier = [True] * (n + 1)
    est_premier[0] = est_premier[1] = False
    for p in range(2, int(n**0.5) + 1):
        if est_premier[p]:
            for multiple in range(p * p, n + 1, p):
                est_premier[multiple] = False
    return [k for k in range(n + 1) if est_premier[k]]

assert crible(20) == [2, 3, 5, 7, 11, 13, 17, 19]
assert crible(1) == []
assert crible(2) == [2]
assert len(crible(100)) == 25
````

### Q1. Quelle est la complexité de `L.pop(0)` sur une liste de n éléments ?

<details><summary>Réponse</summary>

**O(n)** : retirer le premier élément décale les n − 1 autres. Dans une boucle de n tours, le coût total devient **O(n²)**. Pour une file, utiliser `collections.deque` et `popleft()` (O(1)).

</details>


### Q2. En dichotomie, pourquoi `while g <= d` et non `while g < d` ?

<details><summary>Réponse</summary>

Avec `g < d`, on s'arrête quand `g == d` : le dernier élément n'est jamais testé. Il faut `<=` pour examiner la zone réduite à un seul indice. De plus, on met à jour `g = m + 1` et `d = m - 1` pour garantir la terminaison.

</details>


### Q3. Que vaut `A` après ce code, et pourquoi ?

```python
A = [1, 2, 3]
B = A
B[0] = 99
```

<details><summary>Réponse</summary>

`A == [99, 2, 3]` : `B = A` crée un **alias** (même objet). Modifier via `B` modifie `A`. Pour une copie indépendante : `C = A.copy()` ou `A[:]` (copie superficielle), et `copy.deepcopy` pour des listes imbriquées.

</details>


### Q4. Pourquoi le tri fusion est-il stable, et quelle est sa complexité garantie ?

<details><summary>Réponse</summary>

Il est stable grâce à la condition `A[i] <= B[j]` : à égalité, on prend l'élément de gauche en premier, donc l'ordre d'origine des égaux est conservé. Sa complexité est **O(n log n)** dans tous les cas, au prix de **O(n)** de mémoire auxiliaire.

</details>


### Q5. Pourquoi l'exponentiation rapide est-elle en O(log n) ?

<details><summary>Réponse</summary>

À chaque étape on divise l'exposant par 2 (`n //= 2`), donc le nombre d'itérations est le nombre de bits de n, soit **⌊log₂ n⌋ + 1**. La version récursive doit mémoriser `y = puissance(x, n // 2)` pour ne faire qu'**un seul** appel récursif ; sinon on retombe en O(n).

</details>
